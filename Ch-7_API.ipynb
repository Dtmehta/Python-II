## Headers
- headers are passed with requests to provide extra info. , they are optional

```python
headers = {
    "User-Agent" : "Mozilla/5.0",
    "Accept" : "application/json"
}
response = requests.get("https://jsonplaceholder.typicode.com/posts/1" , headers = headers)
print(response.text)
```

### `POST` method
- To create a new resource ; serverside

```python
newpost = {
    "title" : "Learning API",
    "body" : "This is for demo",
    "userId" : 1
}
response = requests.post("https://jsonplaceholder.typicode.com/posts" , json = newpost)
print(response.status_code)
print(response.text)
```

## `PUT` method
- This method is used to update the whole resource

```python
updatepost = {
    "id" :1,
    "title" : "A New Title",
    "body" : "An updated body",
    "uesrId" : 1
}
response = requests.put("https://jsonplaceholder.typicode.com/posts/1" , json = updatepost)
print(response.status_code)
print(response.text)
```

## `PATCH` method
- This method is used to partially update the resource

```python
patched_data = {
    "title" : "A patched title"
}
response = requests.patch("https://jsonplaceholder.typicode.com/posts/1" , json=patched_data)
print(response.status_code)
print(response.text)
```

## `DELETE` method
- this method is delete the resource

```python
response = requests.delete("https://jsonplaceholder.typicode.com/posts/1")
print(response.status_code)
print(response.text)
```

## `PAGINATION`
- This feature allows the API server to control the access of API based on page view

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts"
params = {
    "_page" : 1 ,
    "_limit" : 5
}
response = requests.get(url , params=params)
data = response.json()
print("Posts on page 1:-")
for post in data:
    print(post['id'] , post['title'])
```

```python
# Multiple pages
import requests
url = "https://jsonplaceholder.typicode.com/posts"
page = 1
while page <=3:
    params = {
        "_page" : page ,
        "_limit" : 5
    }
    response = requests.get(url , params=params)
    posts = response.json()
    print(f"\nPage {page} ")
    for post in posts:
        print(post['id'] , post['title'])
    page+=1
```

## RATE LIMIT
- This feature allow the API provider to control the access to the resource

```python
# to check the rate limit of github api
import requests
import pprint
response = requests.get("https://api.github.com/rate_limit")
data = response.json()
pprint.pprint(data)
print("Limit" , data['rate']['limit'])
print("Used" , data['rate']['used'])
print("Left" , data['rate']['remaining'])
print("Reset" , data['rate']['reset'])
```

## Requests multiple times to check ratelimit

```python
import requests
for i in range(40):
    response = requests.get("https://api.github.com/users/octocat")
    print(i+1,response.status_code)
    # In case of exceeding rate limit
    if response.status_code == 403:
        data = response.json()
        if "API rate limit exceeded" in data.get("message",""):
            print("Rate limit exceeded")
            print(data['message'])
            break
```
